#### EDA and data understanding

```python
df.duplicated().sum()
```

```python
df = df.rename(columns={
    "Air temperature [K]": "air_temp", "Process temperature [K]": "process_temp",
    "Rotational speed [rpm]": "rpm", "Torque [Nm]": "torque",
    "Tool wear [min]": "tool_wear", "Machine failure": "failure", "Type": "type"})
```

```python
df
```

```python
df.head()
```

```python
types_names ={'L': 'Low','M': 'Medium', 'H': 'High'}
```

```python
df['type']=df['type'].replace(types_names)
```

```python
df['type']
```

```python
df.head()
```

```python
int(df.isna().sum().sum())
```

```python
df["failure"].value_counts()
```

```python
print(f"\nFailure rate = {df['failure'].mean():.2%}  -> heavily IMBALANCED")
```

```python
print(df["failure"].value_counts())
print("Failure rate: {:.2%}".format(df["failure"].mean()))
sns.countplot(x="failure", data=df)
plt.title("Failure vs No failure")
plt.show()
```

```python
num_cols = ["air_temp", "process_temp", "rpm", "torque", "tool_wear"]

df[num_cols].hist(bins=30, figsize=(12, 7))
plt.tight_layout()
plt.show()
```

#### boxplots to see outliers- outliers exist in rpm and torque

```python
fig, axes = plt.subplots(1, 5, figsize=(16, 4))
for ax, col in zip(axes, num_cols):
    sns.boxplot(y=df[col], ax=ax)
plt.tight_layout()
plt.show()
```
